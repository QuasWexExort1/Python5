```python
import pandas as pd

games = {101: "Witcher 3", 102: "Cyberpunk 2077", 103: "GTA V", 104: "RDR 2", 105: "Elden Ring"}
devs = {101: "CD Projekt Red", 102: "CD Projekt Red", 103: "Rockstar", 104: "Rockstar", 105: "FromSoftware"}
ratings = {101: 9.5, 102: 8.6, 103: 9.2, 104: 9.7, 105: 9.0}

s1 = pd.Series(games, name="Title")
s2 = pd.Series(devs, name="Developer")
s3 = pd.Series(ratings, name="Rating")

s4 = pd.Series({k: f"{s1[k]} от {s2[k]}" for k in s1.index})

max_r = s3.max()
min_r = s3.min()
mean_r = s3.mean()

print(max_r, min_r, mean_r)
```

```python
import pandas as pd
import numpy as np


data_dict = {
    101: {"Title": "Witcher 3", "Year": 2015, "Platform": "PC"},
    102: {"Title": "Cyberpunk 2077", "Year": 2020, "Platform": "PC"},
    103: {"Title": "GTA V", "Year": 2013, "Platform": "PS4"}
}
df1 = pd.DataFrame.from_dict(data_dict, orient='index')


df2 = pd.DataFrame({"Title": s1, "Developer": s2, "Rating": s3})


df3 = pd.concat([df1, df2], axis=1)


print(df3.index.tolist())
print(df3.columns.tolist())
print(df3.values)


print(df3.head(2))
print(df3.iloc[1:4])
print(df3["Title"])
print(df3["Rating"])

target_id = 101
if target_id in df3.index:
    print(df3.loc[target_id, "Title"])

target_dev = "Rockstar"
print(list(df3.loc[df3["Developer"] == target_dev, "Title"]))

```
